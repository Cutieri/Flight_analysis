```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt    
```

```python
df = pd.read_csv(r"E:\Pós\data\nycflights.csv")
```

```python
df
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df
```

```python
#tirando todos os voos cancelados, pois analisaremos atrasos
df = df.dropna(subset=["dep_time", "arr_time"])
```

```python
#formatando dep_time em HH:MM
df = df.copy()

# Converter para string com zeros à esquerda
df["dep_time"] = df["dep_time"].astype(pd.Int64Dtype()).astype(str).str.zfill(4)

# Converter para datetime e formatar como HH:MM
df["dep_time"] = pd.to_datetime(
    df["dep_time"], 
    format="%H%M", 
    errors="coerce"
).dt.strftime("%H:%M")

df
```

```python
df = df.copy()

# Converter para string com zeros à esquerda
df["air_time"] = df["air_time"].astype(pd.Int64Dtype()).astype(str).str.zfill(4)

# Converter para datetime e formatar como HH:MM
df["air_time"] = pd.to_datetime(
    df["air_time"], 
    format="%H%M", 
    errors="coerce"
).dt.strftime("%H:%M")
    
df
```
